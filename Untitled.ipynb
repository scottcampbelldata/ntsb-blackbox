```python
import sqlite3
import pandas as pd

con = sqlite3.connect("data/ntsb.db")
pd.read_sql_query("SELECT COUNT(*) FROM accidents WHERE event_year = 2019", con)
```

```python
pd.read_sql_query("""
    SELECT UPPER(make) AS make_norm, COUNT(*) AS n
    FROM accidents
    GROUP BY UPPER(make)
    ORDER BY n DESC
    LIMIT 10
""", con)
```
